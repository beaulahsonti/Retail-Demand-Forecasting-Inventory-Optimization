```python
import pandas as pd
import os
```

```python
sales_long = pd.read_csv(
    "../data/processed/sales_long_sample.csv"
)
```

```python
calendar = pd.read_csv(
    "../data/processed/calendar_cleaned.csv",
    usecols=[
        "d",
        "date",
        "weekday",
        "month",
        "year",
        "week",
        "day",
        "event_name_1",
        "event_type_1",
        "event_name_2",
        "event_type_2"
    ]
)
```

```python
daily_sales = sales_long.merge(
    calendar,
    on="d",
    how="left"
)
```

```python
daily_sales["date"] = pd.to_datetime(
    daily_sales["date"]
)
```

```python
daily_sales = daily_sales.rename(
    columns={
        "item_id": "item",
        "store_id": "store",
        "dept_id": "department",
        "cat_id": "category",
        "event_name_1": "event"
    }
)
```

```python
daily_sales = daily_sales[
    [
        "id",
        "item",
        "store",
        "department",
        "category",
        "state_id",
        "d",
        "date",
        "sales",
        "event",
        "event_type_1",
        "event_name_2",
        "event_type_2",
        "weekday",
        "month",
        "year",
        "week",
        "day"
    ]
]
```

```python
daily_sales.to_csv(
    "../data/processed/daily_sales_sample.csv",
    index=False
)
```

- Loaded the transformed M5 sales sample.
- Loaded the cleaned M5 calendar dataset.
- Merged sales and calendar data using the M5 `d` identifier.
- Converted M5 daily identifiers into actual calendar dates.
- Added calendar, weekday and event information to the sales records.
