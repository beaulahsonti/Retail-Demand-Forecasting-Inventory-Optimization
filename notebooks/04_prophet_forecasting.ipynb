# 04 — Prophet Demand Forecasting

This notebook runs the Prophet forecasting model and reviews its validation performance.

```python
from pathlib import Path
import pandas as pd
import subprocess
import sys

ROOT = Path.cwd().resolve()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

print("Project root:", ROOT)
```

```python
script = ROOT / "src" / "forecasting" / "prophet_model.py"

result = subprocess.run(
    [sys.executable, str(script)],
    cwd=ROOT,
    text=True,
    capture_output=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError("Prophet forecasting failed.")
```

```python
metrics_path = ROOT / "reports" / "prophet_metrics.csv"

if metrics_path.exists():
    metrics = pd.read_csv(metrics_path)
    display(metrics)
else:
    print("Prophet metrics file was not found.")
```
