# 10 — SQL Analytics with DuckDB

This notebook runs the project's SQL analytics workflow.

Analyses include:
- Demand summary
- Product analysis
- Store analysis
- Forecast analysis
- Inventory analysis

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

print("Project root:", ROOT)
```

```python
script = ROOT / "src" / "analytics" / "run_sql.py"

result = subprocess.run(
    [sys.executable, str(script)],
    cwd=ROOT,
    text=True,
    capture_output=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError("SQL analytics workflow failed.")
```

```python
db_path = ROOT / "data" / "retail_demand.duckdb"

print("DuckDB database exists:", db_path.exists())
print("Database path:", db_path)
```
