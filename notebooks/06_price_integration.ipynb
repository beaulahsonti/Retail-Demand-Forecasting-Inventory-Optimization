```python
import pandas as pd
```

```python
daily_sales = pd.read_csv(
    "../data/processed/daily_sales_sample.csv"
)
```

```python
daily_sales.head()
```

```python
prices = pd.read_csv(
    "../data/raw/sell_prices.csv"
)
```

```python
prices.head()
```

```python
print("Price data shape:", prices.shape)
```

```python
print(prices.columns.tolist())
```

```python
prices = prices.rename(
    columns={
        "item_id": "item",
        "store_id": "store",
        "sell_price": "price"
    }
)
```

```python
prices.head()
```

```python
selected_items = daily_sales["item"].drop_duplicates().head(20).tolist()
```

```python
selected_stores = daily_sales["store"].drop_duplicates().head(5).tolist()
```

```python
print("Selected items:", selected_items)
print("Selected stores:", selected_stores)
```

```python
daily_sales_subset = daily_sales[
    daily_sales["item"].isin(selected_items)
    & daily_sales["store"].isin(selected_stores)
].copy()
```

```python
print(daily_sales_subset.shape)
```

```python
prices_subset = prices[
    prices["item"].isin(selected_items)
    & prices["store"].isin(selected_stores)
].copy()
```

```python
print(prices_subset.shape)
```

```python
print("daily_sales columns:")
print(daily_sales.columns.tolist())
```

```python
print("prices columns:")
print(prices.columns.tolist())
```

```python
calendar = pd.read_csv(
    "../data/processed/calendar_cleaned.csv"
)
```

```python
daily_sales["date"] = pd.to_datetime(daily_sales["date"])
calendar["date"] = pd.to_datetime(calendar["date"])
```

```python
daily_sales = daily_sales.merge(
    calendar[["date", "wm_yr_wk"]],
    on="date",
    how="left"
)
```

```python
print(daily_sales.columns.tolist())
```

```python
print(daily_sales["wm_yr_wk"].isnull().sum())
```

```python
daily_sales_subset = daily_sales[
    daily_sales["item"].isin(selected_items)
    & daily_sales["store"].isin(selected_stores)
].copy()

prices_subset = prices[
    prices["item"].isin(selected_items)
    & prices["store"].isin(selected_stores)
].copy()
```

```python
daily_sales_with_price = daily_sales_subset.merge(
    prices_subset,
    on=["item", "store", "wm_yr_wk"],
    how="left"
)
```

```python
print(daily_sales["wm_yr_wk"].isnull().sum())
print(daily_sales_with_price.shape)
```

```python
daily_sales_with_price.head()
```

```python
print(daily_sales_with_price.shape)
```

```python
print("Missing prices:",
      daily_sales_with_price["price"].isnull().sum())
```

```python
missing_price_pct = (
    daily_sales_with_price["price"].isnull().mean() * 100
)

print("Missing price percentage:",
      round(missing_price_pct, 2), "%")
```

```python
print(
    "Duplicate rows:",
    daily_sales_with_price.duplicated().sum()
)
```

```python
daily_sales_with_price["price"].describe()
```

```python
final_data = daily_sales_with_price[
    [
        "id",
        "item",
        "store",
        "department",
        "category",
        "state_id",
        "date",
        "sales",
        "price",
        "event",
        "event_type_1",
        "event_name_2",
        "event_type_2",
        "weekday",
        "year",
        "month",
        "week",
        "day",
        "wm_yr_wk"
    ]
].copy()
```

```python
final_data.head()
```

```python
output_path = "../data/processed/daily_sales_with_price.csv"

final_data.to_csv(
    output_path,
    index=False
)
```

## Price Integration Summary

- Integrated M5 weekly selling price information with the daily sales dataset.
- Used item, store and week (`wm_yr_wk`) as the merge keys.
- Checked missing price values and duplicate records.
- Created the final analytical dataset containing sales, price, product, store, date and event information.
- Used a selected M5 subset to keep the processed dataset below the project file-size limit.
- Saved the result as `daily_sales_with_price.csv`.
