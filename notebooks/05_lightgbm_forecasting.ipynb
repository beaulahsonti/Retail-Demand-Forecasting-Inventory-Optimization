# 05 — LightGBM Demand Forecasting

This notebook runs the LightGBM forecasting model and reviews:
- Validation metrics
- Feature importance
- Generated predictions

```python
from pathlib import Path
import pandas as pd
import subprocess
import sys

ROOT = Path.cwd().resolve()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

print("Project root:", ROOT)
```

```python
script = ROOT / "src" / "forecasting" / "lightgbm_model.py"

result = subprocess.run(
    [sys.executable, str(script)],
    cwd=ROOT,
    text=True,
    capture_output=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError("LightGBM forecasting failed.")
```

```python
metrics = pd.read_csv(
    ROOT / "reports" / "lightgbm_metrics.csv"
)

importance = pd.read_csv(
    ROOT / "reports" / "lightgbm_feature_importance.csv"
)

display(metrics)

print("Top Feature Importance:")
display(importance.head(15))
```

```python
predictions = pd.read_csv(
    ROOT / "outputs" / "forecasts" / "lightgbm_predictions.csv"
)

print("Forecast rows:", len(predictions))
display(predictions.head())
```
