# 03 — Baseline Demand Forecasting

This notebook runs the project's 7-day lag baseline forecasting model.

## Objectives

- Generate baseline demand predictions
- Evaluate validation performance
- Calculate MAE, RMSE, WAPE and sMAPE
- Review saved baseline predictions

```python
from pathlib import Path
import pandas as pd
import subprocess
import sys

ROOT = Path.cwd().resolve()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

print("Project root:", ROOT)
```

```python
script = ROOT / "src" / "forecasting" / "baseline_model.py"

result = subprocess.run(
    [sys.executable, str(script)],
    cwd=ROOT,
    text=True,
    capture_output=True
)

print(result.stdout)

if result.returncode != 0:
    print(result.stderr)
    raise RuntimeError("Baseline forecasting failed.")
```

```python
metrics_path = ROOT / "reports" / "baseline_metrics.csv"

metrics = pd.read_csv(metrics_path)

display(metrics)
```

```python
prediction_path = ROOT / "outputs" / "forecasts" / "baseline_predictions.csv"

predictions = pd.read_csv(prediction_path)

print("Prediction rows:", len(predictions))
display(predictions.head())
```
