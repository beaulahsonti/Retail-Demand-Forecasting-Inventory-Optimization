# 06 — Forecasting Model Comparison

Comparison of:
- Baseline
- Prophet
- LightGBM

The comparison uses the project's saved validation metrics.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

comparison = pd.read_csv(
    ROOT / "reports" / "model_comparison.csv"
)

display(comparison)
```

```python
metric_columns = [
    "MAE",
    "RMSE",
    "WAPE",
    "sMAPE"
]

available = [
    c for c in metric_columns
    if c in comparison.columns
]

comparison.set_index("model")[available].plot(
    kind="bar",
    figsize=(10, 5)
)

plt.title("Forecasting Model Performance")
plt.ylabel("Metric Value")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
