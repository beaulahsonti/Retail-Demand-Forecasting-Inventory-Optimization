# 07 — Safety Stock Analysis

Review the safety-stock calculations generated by the inventory planning workflow.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().resolve()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

safety_stock = pd.read_csv(
    ROOT / "reports" / "safety_stock.csv"
)

print("Product-store combinations:", len(safety_stock))

display(safety_stock.head())
```

```python
display(safety_stock.describe(include="all").T)
```

```python
if "safety_stock" in safety_stock.columns:

    safety_stock["safety_stock"].plot(
        kind="hist",
        bins=20,
        figsize=(9, 5)
    )

    plt.title("Safety Stock Distribution")
    plt.xlabel("Safety Stock")
    plt.tight_layout()
    plt.show()
```
